Question 18: Find the Maximum Subarray Sum Using Kadane's Algorithm.

Prompt: Write a Python program to find the maximum possible sum of a non-empty contiguous subarray using Kadane's algorithm.

Kadane's Algorithm.

Definition: Kadane's Algorithm is used to find the maximum sum of a contiguous subarray in a given array.
Working:
1. Initialize current_sum and max_sum with the first element.
2. Traverse the array from the second element.
3. At each step, decide whether to start a new subarray or continue the existing one.
4. Update max_sum with the largest sum found.

```python

numbers = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

current_sum = numbers[0]
max_sum = numbers[0]

for num in numbers[1:]:
    current_sum = max(num, current_sum + num)
    max_sum = max(max_sum, current_sum)

print("Maximum Subarray Sum:", max_sum)
```

Question 24: Implement Binary Search.

Prompt: Write a Python program to search for a target element in a sorted list using the Binary Search algorithm. Return the index of the target if found; otherwise, return -1.

```python
numbers = [10, 20, 30, 40, 50, 60, 70]
target = 40

left = 0
right = len(numbers) - 1
result = -1

while left <= right:
    mid = (left + right) // 2

    if numbers[mid] == target:
        result = mid
        break
    elif numbers[mid] < target:
        left = mid + 1
    else:
        right = mid - 1

print("Element Found at Index:", result)
```

Binary Search works on sorted data and has a time complexity of O(log n).

Question 25: Implement Bubble Sort.

Prompt: Write a Python program to sort a list of integers in ascending order using the Bubble Sort algorithm without using built-in sorting functions.

```python
numbers = [64, 34, 25, 12, 22, 11, 90]

n = len(numbers)

for i in range(n):
    swapped = False

    for j in range(0, n - i - 1):
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
            swapped = True

    if not swapped:
        break

print("Sorted List:", numbers)
```

Bubble Sort repeatedly compares adjacent elements and swaps them if they are in the wrong order.


Question 26: Implement Selection Sort.

Prompt: Write a Python program to sort a list of integers in ascending order using Selection Sort without using built-in sorting functions.

```python
numbers = [64, 25, 12, 22, 11]

n = len(numbers)

for i in range(n):
    min_index = i

    for j in range(i + 1, n):
        if numbers[j] < numbers[min_index]:
            min_index = j

    numbers[i], numbers[min_index] = (
        numbers[min_index], numbers[i]
    )

print("Sorted List:", numbers)
```

Selection Sort finds the smallest element in the unsorted portion and places it in its correct position.

Question 27: Implement Insertion Sort.

Prompt: Write a Python program to sort a list of integers in ascending order using Insertion Sort without using built-in sorting functions.

```python
numbers = [12, 11, 13, 5, 6]

for i in range(1, len(numbers)):
    key = numbers[i]
    j = i - 1

    while j >= 0 and numbers[j] > key:
        numbers[j + 1] = numbers[j]
        j -= 1

    numbers[j + 1] = key

print("Sorted List:", numbers)
```

Insertion Sort builds a sorted portion of the list one element at a time. It performs well on small or nearly sorted datasets.

Question 28: Implement Merge Sort.

Prompt: Write a Python program to sort a list of integers in ascending order using the Merge Sort algorithm based on the divide-and-conquer technique.

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


numbers = [38, 27, 43, 3, 9, 82, 10]

sorted_numbers = merge_sort(numbers)

print("Sorted List:", sorted_numbers)
```

Merge Sort divides the array into smaller parts, sorts them recursively, and merges them into a sorted array.

```python

def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


num = 5

print("Factorial of", num, "is:", factorial(num))
```

Question 32: Generate Pascal's Triangle.

Prompt: Write a Python program to generate the first N rows of Pascal's Triangle using loops.
In Pascal's Triangle, every interior number is the sum of the two numbers directly above it.

```python
n = 5

triangle = []

for i in range(n):
    row = [1] * (i + 1)

    for j in range(1, i):
        row[j] = triangle[i - 1][j - 1] + triangle[i - 1][j]

    triangle.append(row)

print("Pascal's Triangle:")

for row in triangle:
    print(" ".join(map(str, row)))
```

Pascal's Triangle.

Definition: Pascal's Triangle is a triangular arrangement of numbers where each number is the sum of the two numbers directly above it.

Definition: Dynamic Programming is an algorithmic technique used to solve complex problems by breaking them into smaller overlapping subproblems and storing their results to avoid repeated calculations.

Two Main Approaches:
1. Memoization (Top-Down): Uses recursion and stores previously calculated results.
2. Tabulation (Bottom-Up): Uses iteration and a table to store results.

Question 35: Find the Next Greater Element Using a Stack.

Prompt: Given a list of integers, write a Python program to find the next greater element for each element. The next greater element is the first element to its right that is greater than the current element. If no greater element exists, return -1.

```python
numbers = [4, 5, 2, 25]

result = [-1] * len(numbers)
stack = []

for i in range(len(numbers)):
    while stack and numbers[i] > numbers[stack[-1]]:
        index = stack.pop()
        result[index] = numbers[i]

    stack.append(i)

print("Original List:", numbers)
print("Next Greater Elements:", result)
```

This problem uses a monotonic stack, which stores indices whose next greater elements have not yet been found.
Time Complexity: O(n)
Space Complexity: O(n)

Question 36: Evaluate a Postfix Expression Using a Stack.

Prompt: Write a Python program to evaluate a valid postfix expression using a stack. In postfix notation, operators appear after their operands. Assume the expression contains space-separated integer operands and the operators +, -, *, and /.

```python
def evaluate_postfix(expression):
    stack = []

    for token in expression.split():
        if token not in {"+", "-", "*", "/"}:
            stack.append(int(token))
        else:
            b = stack.pop()
            a = stack.pop()

            if token == "+":
                stack.append(a + b)
            elif token == "-":
                stack.append(a - b)
            elif token == "*":
                stack.append(a * b)
            elif token == "/":
                stack.append(a / b)

    return stack.pop()


expression = "5 3 2 * +"

result = evaluate_postfix(expression)

print("Postfix Expression:", expression)
print("Result:", result)
```

Question 37: Implement a Queue Using Python deque.

Prompt: Write a Python program to implement a queue using the collections.deque module. Perform enqueue, dequeue, and front operations following the First-In, First-Out (FIFO) principle.

```python
from collections import deque

queue = deque()

# Enqueue elements
queue.append(10)
queue.append(20)
queue.append(30)

print("Queue:", list(queue))

# Dequeue element
removed = queue.popleft()
print("Dequeued Element:", removed)

# Front element
print("Front Element:", queue[0])

print("Queue After Dequeue:", list(queue))
```
