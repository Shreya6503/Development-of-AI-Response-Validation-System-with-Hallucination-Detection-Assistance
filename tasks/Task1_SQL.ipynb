Question 14: Count Employees in Each Department.

Prompt: Write an SQL query to display each department name and the number of employees working in that department using GROUP BY.

```python
run_query("""
SELECT
    d.dept_name,
    COUNT(e.emp_id) AS employee_count
FROM Departments d
LEFT JOIN Employees e
    ON d.dept_id = e.department_id
GROUP BY d.dept_id, d.dept_name
ORDER BY d.dept_id;
""")
```

GROUP BY combines rows having the same values into groups. Here, LEFT JOIN also ensures that departments with zero employees would appear with a count of 0.


Question 15: Find Departments Having an Average Salary Greater Than 55,000.

Prompt: Write an SQL query to display departments whose average employee salary is greater than 55,000 using GROUP BY and HAVING.

```python
run_query("""

SELECT
    d.dept_name,
    AVG(e.salary) AS average_salary
FROM Employees e
JOIN Departments d
    ON e.department_id = d.dept_id
GROUP BY d.dept_id, d.dept_name
HAVING AVG(e.salary) > 55000
ORDER BY average_salary DESC;
""")
```

WHERE filters individual rows before grouping, while HAVING filters groups after aggregation.

Question 19: Display All Employees and Departments Using FULL OUTER JOIN.

Prompt: Write an SQL query to display all employees and all departments, including unmatched records from either table, using the concept of FULL OUTER JOIN.

```python
run_query("""
SELECT
    e.emp_name,
    d.dept_name
FROM Employees e
LEFT JOIN Departments d
    ON e.department_id = d.dept_id

UNION ALL

SELECT
    e.emp_name,
    d.dept_name
FROM Departments d
LEFT JOIN Employees e
    ON d.dept_id = e.department_id
WHERE e.emp_id IS NULL

ORDER BY dept_name, emp_name;
""")
```

FULL OUTER JOIN returns matching rows and unmatched rows from both tables.

Question 20: Find Employees Working in the Same Department Using SELF JOIN.

Prompt: Write an SQL query to find pairs of employees who work in the same department. Avoid pairing an employee with themselves and avoid duplicate reversed pairs.

```python
run_query("""
SELECT
    e1.emp_name AS employee_1,
    e2.emp_name AS employee_2,
    d.dept_name
FROM Employees e1
INNER JOIN Employees e2
    ON e1.department_id = e2.department_id
    AND e1.emp_id < e2.emp_id
INNER JOIN Departments d
    ON e1.department_id = d.dept_id
ORDER BY d.dept_id, e1.emp_id, e2.emp_id;
""")
```

Question 24: Find Employees Earning More Than Their Department's Average Salary.

Prompt: Write an SQL query to find employees whose salaries are greater than the average salary of their respective departments using a correlated subquery.

```python
run_query("""
SELECT
    e.emp_name,
    e.salary,
    d.dept_name
FROM Employees e
JOIN Departments d
    ON e.department_id = d.dept_id
WHERE e.salary > (
    SELECT AVG(e2.salary)
    FROM Employees e2
    WHERE e2.department_id = e.department_id
)
ORDER BY d.dept_id, e.emp_id;
""")
```

Question 25: Find Departments Having at Least One Employee Earning More Than 70,000.

Prompt: Write an SQL query to find department names where at least one employee earns more than 70,000 using the EXISTS operator.

```python
run_query("""
SELECT
    d.dept_name
FROM Departments d
WHERE EXISTS (
    SELECT 1
    FROM Employees e
    WHERE e.department_id = d.dept_id
      AND e.salary > 70000
)
ORDER BY d.dept_id;
""")
```

EXISTS returns true when the subquery produces at least one row. It is useful for checking whether related records exist without returning duplicates.

What Are SQL Window Functions?

Window functions perform calculations across related rows without combining them into a single row, unlike GROUP BY.

The general syntax is:

FUNCTION_NAME()OVER (

    PARTITION BY column_name
    ORDER BY column_name
)

OVER() defines the window.

PARTITION BY divides rows into groups.

ORDER BY determines the order within each window.

Question 27: Rank Employees by Salary Using RANK().

Prompt: Write an SQL query to rank employees according to their salaries in descending order using RANK(). Employees with equal salaries should receive the same rank.

```python
run_query("""
SELECT
    emp_name,
    salary,
    RANK() OVER (
        ORDER BY salary DESC
    ) AS salary_rank
FROM Employees
ORDER BY salary DESC, emp_id;
""")
```

RANK() assigns the same rank to equal values but skips subsequent ranks. Because two employees have rank 2, the next employee receives rank 4.

Question 28: Rank Employees Using DENSE_RANK().

Prompt: Write an SQL query to assign salary ranks using DENSE_RANK() without skipping ranks when employees have equal salaries.

```python
run_query("""
SELECT
    emp_name,
    salary,
    DENSE_RANK() OVER (
        ORDER BY salary DESC
    ) AS dense_salary_rank
FROM Employees
ORDER BY salary DESC, emp_id;
""")
```

DENSE_RANK() gives equal values the same rank without gaps. It is useful for finding the second-highest distinct salary.

Question 29: Calculate the Running Total of Employee Salaries.

Prompt: Write an SQL query to calculate the cumulative salary total for employees ordered by employee ID using SUM() as a window function.

```python
run_query("""
SELECT
    emp_name,
    salary,
    SUM(salary) OVER (
        ORDER BY emp_id
        ROWS BETWEEN UNBOUNDED PRECEDING
        AND CURRENT ROW
    ) AS running_total
FROM Employees
ORDER BY emp_id;
""")
```

A running total adds the current row's value to the sum of all preceding rows. ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW explicitly defines the cumulative window.

Question 30: Find the Highest-Paid Employee in Each Department.

Prompt: Write an SQL query to find the highest-paid employee in each department using DENSE_RANK() and PARTITION BY. Include all employees tied for the highest salary.

```python
run_query("""
WITH RankedEmployees AS (
    SELECT
        e.emp_name,
        d.dept_name,
        e.salary,
        d.dept_id,
        e.emp_id,
        DENSE_RANK() OVER (
            PARTITION BY e.department_id
            ORDER BY e.salary DESC
        ) AS salary_rank
    FROM Employees e
    JOIN Departments d
        ON e.department_id = d.dept_id
)
SELECT
    emp_name,
    dept_name,
    salary
FROM RankedEmployees
WHERE salary_rank = 1
ORDER BY dept_id, emp_id;
""")
```

PARTITION BY restarts the ranking within each department. DENSE_RANK() = 1 returns the highest-paid employee or employees from each department, including ties.

Question 35: Replace NULL Values Using COALESCE().

Prompt: Write an SQL query to display employee names and their department names. If an employee has no matching department, display "Not Assigned" instead of NULL using COALESCE().

```python
run_query("""
SELECT
    e.emp_name,
    COALESCE(d.dept_name, 'Not Assigned') AS department
FROM Employees e
LEFT JOIN Departments d
    ON e.department_id = d.dept_id
ORDER BY e.emp_id;
""")
```

COALESCE() returns the first non-NULL value from its arguments.
Because all employees in the sample data have assigned departments, the output contains no "Not Assigned" values. However, if an employee's department_id is NULL or has no matching department, the query will display "Not Assigned".
